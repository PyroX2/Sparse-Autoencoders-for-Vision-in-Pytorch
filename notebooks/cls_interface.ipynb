```python
# Load classification config
config = read_cls_config()

general_config = read_general_config()
model_config = read_model_config()  # mlp/cnn/vit

project_name = general_config.project_name
dataset = config.dataset  # coloredmnist/waterbirds/celeba
lr = config.lr
batch_size = config.batch_size
epochs = config.epochs
sae_input_layer = (
    config.sae_input_layer
)  # Layer which activations will be used as inputs for training SAE
```

```python
model = ClsModel.from_config(model_config.cls_model)
loss = nn.LossFunction()
optimizer = torch.optim.Optimizer(model.parameters(), lr)

trainer = ClsTrainer(dataset, model, loss, optimizer)

trainer.train(batch_size, epochs)
trainer.evaluate()
```
