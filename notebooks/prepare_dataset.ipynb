```python
import torch
from torchvision import datasets
from torchvision.transforms import v2
import matplotlib.pyplot as plt
from sklearn import model_selection
from config import SEED, VAL_SIZE, COLORED_PROPORTION, COLORS
from dataset import ColoredMNISTDataset
```

```python
plt.style.use("default")
```

```python
train_transforms = v2.Compose(
	[
        v2.ToImage(),
        v2.ToDtype(torch.float32, scale=True)
	]
)
test_transforms = v2.Compose(
	[
        v2.ToImage(),
        v2.ToDtype(torch.float32, scale=True)
	]
)
```

```python
train_dataset = datasets.MNIST(root="./data", download=True)
test_dataset = datasets.MNIST(root="./data", train=False, download=True)
```

```python
### Display proportions before train/val split
labels = torch.tensor([y for x, y in train_dataset])

plt.hist(labels, bins=10)
plt.title("Train Dataset Label Distribution before splitting")
```

```python
train_split, val_split = model_selection.train_test_split(train_dataset, test_size=VAL_SIZE, stratify=labels, random_state=SEED)
```

```python
# Display proportions after train/val split
train_labels = torch.tensor([y for x, y in train_split])
val_labels = torch.tensor([y for x, y in val_split])

plt.hist(train_labels, bins=10)
plt.title("Train Dataset Label Distribution")
plt.show()

plt.hist(val_labels, bins=10)
plt.title("Val Dataset Label Distribution")
plt.show()
```

```python
train_dataset = ColoredMNISTDataset(train_split, colored_proportions=COLORED_PROPORTION)
val_dataset = ColoredMNISTDataset(val_split, colored_proportions=COLORED_PROPORTION)
clean_val_dataset = ColoredMNISTDataset(val_split, colored_proportions=0)
fully_colored_val_dataset = ColoredMNISTDataset(val_split, colored_proportions=1)
colors_flipped_val_dataset = ColoredMNISTDataset(val_split, colored_proportions=1, colors_list=COLORS[::-1])
```

```python
torch.save(train_dataset, "./data/train_dataset.pt")
torch.save(val_dataset, "./data/val_dataset.pt")
torch.save(clean_val_dataset, "./data/clean_val_dataset.pt")
torch.save(fully_colored_val_dataset, "./data/fully_colored_val_dataset.pt")
torch.save(colors_flipped_val_dataset, "./data/colors_flipped_val_dataset.pt")
```
